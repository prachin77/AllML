```python
import numpy as np
```

### Creating array from list

```python
# 1D array
arr_1d = np.array([1,2,3,4,5])
print(f"1D array : {arr_1d}")

arr_2d = np.array([[1,2] , [1,2]])
print(f"2D array : {arr_2d}")
```

### List vs np array

```python
l = [1,2,3]
arr = np.array([1,2,3])
arr1 = np.array([4,5,6])

print(f"normal list multiplication = {l * 2}")
print(f"numpy matrice list multiplication = {arr * 2}")
print(f"two matrices multiplication = {arr * arr1}")
```

```python
# Multiplying n-dim matrices
# @ = Matrix multiplication
# * = normal element wise multiplication

arr2 = np.array([[1,2,3]])
arr3 = np.array([[4,5,6]])

print(f"noraml multiplication = {arr2 * arr3}")

arr2 = np.array([[1,2,3]])
arr3 = np.array([
        [4,5,6],
        [1,2,3],
        [7,8,9]
    ])

print(f"matrix multiplication = {arr2 @ arr3}")
```

### Creating matrices using numpy

```python
zeros = np.zeros((2,2))
ones = np.ones((2,2))
full = np.full((2,2) , 69)

print(f"zeroes matrix = \n{zeros}")
print(f"ones matrix = \n{ones}")
print(f"full matrix = \n{full}")
```

```python
# Random is a class in itself
# The major diff between these 2 is how you pass ur dim
random = np.random.random((2,2)) # takes tuple
rand = np.random.rand(2,2) # takes seperate value
randint = np.random.randint(1,100,(2,10))

print(f"random matrix = \n{random}")
print(f"rand matrix = \n{rand}")
print(f"randint matrix = \n{randint}")
```

### Vector , Matrix and Tensor

```python
v = np.array([1,2,3])
m = np.array([[1,2] , [2,3]])

# Tensor = more than 2 dimension or multiple dimension
tensor = np.array(
    [
        [
            [1,2],
            [3,4]
        ] ,
        [
            [5,6],
            [7,8]
        ]
    ]
)

print(f"vector = \n{v}")
print(f"matrix = \n{m} of dim = {m.shape}")
print(f"Tensor = \n{tensor}")
# basic properties
print(f"Tensor shape = {tensor.shape}")
print(f"Tensor dimension = {tensor.ndim}")
print(f"Tensor size = {tensor.size}") # total no of elements inside the array
print(tensor[0])
```

### Array Reshaping

```python
a = np.arange(1,11)

# always remember that while reshaping dim u pass in the parameter ,
# their multiplication should be equal to the number of elements
# inside the array u r reshaping .
# Like here -> 2 * 5 = 10
reshape = a.reshape((2,5))
flattened = reshape.flatten()
transpose = reshape.T # rows to col and vice-versa

print(f"Reshaped array = \n{reshape}")
print(f"Flattened array = \n{flattened}")
print(f"Transposed array = \n{transpose}")
```
